Phase 3: DATA CLEANING 

```python
#1. Importing Libraries
import pandas as pd
import numpy as np 

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
#2. Load the Dataset
df = pd.read_csv("final_reviews_dataset.csv")
print("Dataset Loaded Successfully!")
df.head()
```

```python
#3. dataset Information
print("Shape: " , df.shape)
df.info()
df.describe(include="all")
```

```python
#4: Check Missing Values
df.isnull().sum()
```

```python
#5: Check Duplicate Reviews
duplicates = df.duplicated(subset=["review"]).sum()
print("duplicate Reviews: ", duplicates)
```

```python
#6: Remove Empty Reviews 
df.dropna(subset=["review"], inplace = True)

df = df[df["review"].str.strip() != ""]

df.isnull().sum()
```

```python
#7: Platform Names 

df["platform"] = df["platform"].str.strip().str.title()
df["platform"].value_counts()
```

```python
#8: Source Names
df["source"] = df["source"].str.strip().str.title()
df["source"].value_counts()
```

```python
#9: Convert Ratings 
df["rating"] = pd.to_numeric(df["rating"], errors = "coerce")
df["rating"] = df["rating"].fillna(0)
df["rating"] = df["rating"].astype(int)
```

```python
df["rating"].value_counts().sort_index()
```

```python
#10: Convert date

df["date"] = pd.to_datetime(df["date"], errors="coerce")

df["date"].head()
```

```python
#11: Remove Invalid Rating

df = df[(df["rating"] >= 1) & (df["rating"] <= 5)]

df["rating"].value_counts().sort_index()
```

```python
#12 Review length

df["review_length"] = df["review"].apply(len)

df["review_length"].describe()
```

```python
#13: Remove Very Short Reviews 

df = df[df["review_length"] > 5]
df.shape
```

```python
#14: Final Data Summery 
print("Final Shape: ", df.shape)
df.head()
```

```python
#15: saving the final dataset

df.to_csv(
    "clean_reviews_dataset.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Clean dataset saved successfully!")
```
